## 2.2. Outlier Treatment & Verification

```python
def cap_outliers_iqr(data, col, factor=1.5):
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - factor * IQR
    upper = Q3 + factor * IQR
    data[col] = data[col].clip(lower, upper)
    return data

df_capped = df.copy()
for col in ['sales', 'profit', 'shipping_cost']:
    df_capped = cap_outliers_iqr(df_capped, col)

print("Trước capping - sales max:", df['sales'].max())
print("Sau capping   - sales max:", df_capped['sales'].max())
```

```python
df['profit_flag'] = (df['profit'] > 0).astype(int)
df_capped['profit_flag'] = (df_capped['profit'] > 0).astype(int)

print(df['profit_flag'].value_counts(normalize=True))
```

# 3. Model Building

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV, learning_curve

def evaluate_regression(model_name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    print(f"[{model_name}] MAE={mae:.2f} | RMSE={rmse:.2f} | R2={r2:.4f}")
    return {'model': model_name, 'MAE': mae, 'RMSE': rmse, 'R2': r2}
```

```python
def plot_learning_curve(estimator, X, y, title, cv=5, scoring='neg_root_mean_squared_error'):
    train_sizes, train_scores, val_scores = learning_curve(
        estimator, X, y, cv=cv, scoring=scoring,
        train_sizes=np.linspace(0.1, 1.0, 6), n_jobs=N_JOBS_OUTER, random_state=42
    )

    is_loss = 'neg' in scoring   # True: RMSE/MAE (thấp = tốt) | False: F1/Accuracy (cao = tốt)
    train_adj = -train_scores if is_loss else train_scores
    val_adj = -val_scores if is_loss else val_scores

    train_mean, train_std = train_adj.mean(axis=1), train_adj.std(axis=1)
    val_mean, val_std = val_adj.mean(axis=1), val_adj.std(axis=1)

    metric_name = 'RMSE' if 'squared_error' in scoring else scoring.upper()
    kind = 'Error' if is_loss else 'Score'          # Fix #2: phân biệt loss vs score trong nhãn
    max_pct = train_sizes[-1] / len(X) * 100

    # Fix #4: rút gọn cách hiển thị cv - tránh in nguyên repr dài của StratifiedKFold(...)
    if isinstance(cv, int):
        cv_label = f"cv={cv}"
    else:
        n_splits = cv.get_n_splits()
        cv_kind = 'phân tầng' if 'Stratified' in cv.__class__.__name__ else 'thường'
        cv_label = f"cv={n_splits} (K-Fold {cv_kind})"

    print(f"[{title}] {cv_label} -> điểm cuối đạt {train_sizes[-1]}/{len(X)} dòng train ({max_pct:.1f}%).")

    plt.figure(figsize=(8, 5))
    plt.plot(train_sizes, train_mean, 'o-', color='blue', label=f'Train {kind} ({metric_name})')
    plt.fill_between(train_sizes, train_mean - train_std, train_mean + train_std,   # Fix #3: dải ±std
                      alpha=0.15, color='blue')
    plt.plot(train_sizes, val_mean, 'o-', color='orange', label=f'Validation {kind} ({metric_name})')
    plt.fill_between(train_sizes, val_mean - val_std, val_mean + val_std, alpha=0.15, color='orange')
    plt.title(f'Learning Curve - {title}')
    plt.xlabel(f'Số lượng mẫu huấn luyện (tối đa {max_pct:.0f}% của {len(X)} dòng, {cv_label})')
    plt.ylabel(metric_name)   # Fix #1: nhãn đúng "RMSE"/"F1", không còn in chuỗi scoring thô
    plt.legend()
    plt.grid(True)
    plt.tight_layout()   # Fix #5: tránh nhãn trục dài bị tràn/cắt ra ngoài khung hình
    plt.show()
    return train_sizes, train_mean, train_std, val_mean, val_std

results = []
N_JOBS_OUTER = 1 
N_JOBS_INNER = 1
```

```python
# Learning curve bổ sung: theo SỐ CÂY, luôn dùng 100% dữ liệu train
from sklearn.base import clone
import warnings
warnings.filterwarnings('ignore', message='.*class_weight presets.*warm_start.*')

def plot_tree_growth_curve(model_class, base_params, preprocessor, X_tr, y_tr, X_te, y_te,
                            title, max_n_estimators, step=15, task='regression'):
    prep = clone(preprocessor)
    Xtr_enc = prep.fit_transform(X_tr, y_tr)
    Xte_enc = prep.transform(X_te)
    steps = list(range(step, max_n_estimators + 1, step))
    if steps[-1] != max_n_estimators:
        steps.append(max_n_estimators)
    model = model_class(warm_start=True, random_state=42, n_jobs=N_JOBS_INNER, **base_params)
    train_scores, val_scores = [], []
    for n in steps:
        model.set_params(n_estimators=n)
        model.fit(Xtr_enc, y_tr)
        if task == 'regression':
            train_scores.append(np.sqrt(mean_squared_error(y_tr, model.predict(Xtr_enc))))
            val_scores.append(np.sqrt(mean_squared_error(y_te, model.predict(Xte_enc))))
        else:
            train_scores.append(f1_score(y_tr, model.predict(Xtr_enc)))
            val_scores.append(f1_score(y_te, model.predict(Xte_enc)))
        print(f"  n_estimators={n:>3} | dùng {Xtr_enc.shape[0]} dòng train (100%) "
              f"| Train={train_scores[-1]:.4f} | Test={val_scores[-1]:.4f}")
    plt.figure(figsize=(8, 5))
    plt.plot(steps, train_scores, 'o-', color='blue', label='Train score')
    plt.plot(steps, val_scores, 'o-', color='orange', label='Validation (test) score')
    plt.axvline(x=max_n_estimators, color='red', linestyle='--', label=f'Số cây đã chọn = {max_n_estimators}')
    plt.title(f'Learning Curve theo số cây - {title}')
    plt.xlabel(f'Số cây (n_estimators) — mỗi điểm đều dùng 100% {Xtr_enc.shape[0]} dòng train')
    plt.ylabel('RMSE' if task == 'regression' else 'F1-score')
    plt.legend()
    plt.grid(True)
    plt.show()
    return steps, train_scores, val_scores
```

### 3.2.2. Mô hình xếp hạng hiệu quả khu vực - Đặc trưng TRỄ (Lag Features)

```python
from sklearn.metrics import accuracy_score
df['order_quarter_label'] = df['order_date'].dt.to_period('Q').astype(str)

rq_agg = df.groupby(['region', 'order_quarter_label']).agg(
    avg_profit_margin=('profit_margin', 'mean'),
    profit_rate=('profit_flag', 'mean'),
    avg_sales=('sales', 'mean'),
    avg_discount=('discount', 'mean'),
    avg_shipping_cost=('shipping_cost', 'mean'),
    avg_quantity=('quantity', 'mean'),
    total_orders=('order_id', 'nunique')
).reset_index()

top_cat_q = (df.groupby(['region', 'order_quarter_label', 'category']).size() /
             df.groupby(['region', 'order_quarter_label']).size()).unstack(fill_value=0)
top_cat_q.columns = [f'share_{c}' for c in top_cat_q.columns]
rq_agg = rq_agg.merge(top_cat_q, on=['region', 'order_quarter_label'])
rq_agg['year'] = rq_agg['order_quarter_label'].str[:4].astype(int)

# Ngưỡng phân nhãn chỉ được học từ giai đoạn huấn luyện
train_margin = rq_agg.loc[rq_agg['year'] < 2014, 'avg_profit_margin']
q33, q66 = train_margin.quantile([1/3, 2/3])

rq_agg['efficiency_label'] = pd.cut(
    rq_agg['avg_profit_margin'],
    bins=[-np.inf, q33, q66, np.inf],
    labels=['Thấp', 'Trung bình', 'Cao']
)
print(f"Ngưỡng phân nhãn học từ 2011-2013: q33={q33:.4f}, q66={q66:.4f}")

# --- Đặc trưng trễ: trung bình trượt 4 quý LIỀN TRƯỚC (loại bỏ chính quý đang gán nhãn) ---
rq_sorted = rq_agg.sort_values(['region', 'order_quarter_label']).reset_index(drop=True)

lag_source_cols = ['avg_sales', 'avg_discount', 'avg_shipping_cost', 'avg_quantity', 'total_orders'] + \
                   [c for c in rq_sorted.columns if c.startswith('share_')]

for col in lag_source_cols:
    rq_sorted[f'{col}_lag'] = (
        rq_sorted.groupby('region')[col]
                 .transform(lambda s: s.shift(1).rolling(4, min_periods=1).mean())
    )

feature_cols_lag = [f'{c}_lag' for c in lag_source_cols]
rq_lag = rq_sorted.dropna(subset=feature_cols_lag).copy()  # bỏ quý đầu/khu vực (13 dòng, chưa có lịch sử)
print(f"Số dòng dùng được: {rq_lag.shape[0]} (đã bỏ {rq_sorted.shape[0]-rq_lag.shape[0]} dòng chưa có lịch sử)")

X_lag = rq_lag[feature_cols_lag]
y_lag = rq_lag['efficiency_label']
train_mask = rq_lag['year'] < 2014
test_mask  = rq_lag['year'] == 2014

print("\nPhân bố nhãn thực tế trên tập kiểm định 2014:")
print(y_lag[test_mask].value_counts())

rf_region_lag = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42, n_jobs=N_JOBS_INNER)
rf_region_lag.fit(X_lag[train_mask], y_lag[train_mask])
y_pred_lag = rf_region_lag.predict(X_lag[test_mask])

print("\n=== Đặc trưng trễ (trượt 4 quý), train 2011-2013 -> test 2014 ===")
print(classification_report(y_lag[test_mask], y_pred_lag, zero_division=0))
lag_accuracy = accuracy_score(y_lag[test_mask], y_pred_lag)
print("Accuracy:", lag_accuracy)
```

## 3.3. K-Means Clustering (RFM)

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

snapshot_date = df['order_date'].max() + pd.Timedelta(days=1)
rfm = df.groupby('customer_name').agg(
    Recency=('order_date', lambda x: (snapshot_date - x.max()).days),
    Frequency=('order_id', 'nunique'),
    Monetary=('sales', 'sum')
).reset_index()

print("Độ lệch (skewness) từng biến RFM:")
print(rfm[['Recency','Frequency','Monetary']].skew())
rfm.head()
```

```python
rfm['Recency_log'] = np.log1p(rfm['Recency'])
rfm['Monetary_log'] = np.log1p(rfm['Monetary'])

print("Độ lệch phân phối (skewness) sau log-transform:")
print(rfm[['Recency_log','Frequency','Monetary_log']].skew())

fig, axes = plt.subplots(2, 2, figsize=(12,7))
sns.histplot(rfm['Recency'], kde=True, ax=axes[0,0]).set_title('Recency - Trước log')
sns.histplot(rfm['Recency_log'], kde=True, ax=axes[0,1]).set_title('Recency - Sau log')
sns.histplot(rfm['Monetary'], kde=True, ax=axes[1,0]).set_title('Monetary - Trước log')
sns.histplot(rfm['Monetary_log'], kde=True, ax=axes[1,1]).set_title('Monetary - Sau log')
plt.tight_layout()
plt.show()
```

```python
scaler_rfm = StandardScaler()
rfm_scaled = scaler_rfm.fit_transform(rfm[['Recency_log','Frequency','Monetary_log']])

# Elbow Method - chọn số cụm k tối ưu
inertias = []
sil_scores = []
K_range = range(2, 9)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(rfm_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(rfm_scaled, labels))

fig, axes = plt.subplots(1, 2, figsize=(13,5))
axes[0].plot(K_range, inertias, 'o-')
axes[0].set_title('Elbow Method - Phương pháp khuỷu tay')
axes[0].set_xlabel('Number of clusters k - Số cụm k')
axes[0].set_ylabel('Inertia - Tổng bình phương khoảng cách')

axes[1].plot(K_range, sil_scores, 'o-', color='green')
axes[1].set_title('Silhouette Score theo k')
axes[1].set_xlabel('Number of clusters k - Số cụm k')
axes[1].set_ylabel('Silhouette Score')
plt.tight_layout()
plt.show()
```

```python
def predict_sales(order_dict, model=final_model, reference_columns=X_full.columns):

    order_df = pd.DataFrame([order_dict])

    missing = set(reference_columns) - set(order_df.columns)
    if missing:
        raise ValueError(f"Thiếu các cột bắt buộc để dự đoán: {missing}")

    extra = set(order_df.columns) - set(reference_columns)
    if extra:
        print(f"Cảnh báo: bỏ qua các cột không dùng để huấn luyện: {extra}")

    order_df = order_df[reference_columns]

    return model.predict(order_df)[0]

print("Đã định nghĩa hàm predict_sales().")
```

```python
catboost_rmse = results_df[results_df['model'] == 'CatBoost Regressor']['RMSE'].values[0]
catboost_r2 = results_df[results_df['model'] == 'CatBoost Regressor']['R2'].values[0]
clf_auc = roc_auc_score(y_test_clf, y_proba_rfc)
silhouette_k4 = silhouette_score(rfm_scaled, rfm['Cluster'])
region_loo_accuracy = (y_region == y_pred_region).mean()

print(f"CatBoost: R²={catboost_r2:.3f}, RMSE={catboost_rmse:.0f}")
print(f"Classifier: AUC={clf_auc:.3f}")
print(f"K-Means: Silhouette={silhouette_k4:.3f}")
print(f"Region Classifier (LOO): Accuracy={region_loo_accuracy:.0%}")


def generate_cluster_labels(cluster_profile):
    remaining = list(cluster_profile.index)
    labels = {}

    newest = cluster_profile.loc[remaining, 'Recency'].idxmin()
    labels[newest] = "Khách hàng mới / đang hoạt động tích cực"
    remaining.remove(newest)

    vip = cluster_profile.loc[remaining, 'Monetary'].idxmax()
    labels[vip] = "Khách hàng trung thành / giá trị cao (VIP)"
    remaining.remove(vip)

    at_risk = cluster_profile.loc[remaining, 'Recency'].idxmax()
    labels[at_risk] = "Khách hàng có nguy cơ rời bỏ"
    remaining.remove(at_risk)

    for c in remaining:
        labels[c] = "Khách hàng ít gắn kết"
    return labels


CLUSTER_LABELS = generate_cluster_labels(cluster_profile)
print("Nhãn cụm:", CLUSTER_LABELS)

def plot_sensitivity_chart(base_order_dict):
    """Vẽ 2 biểu đồ: doanh thu & xác suất có lãi biến thiên ra sao khi
    thay đổi Discount hoặc Shipping Cost, giữ nguyên các trường còn lại."""
    fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

    discount_range = np.linspace(0, 0.8, 17)
    sales_d, prob_d = [], []
    for d in discount_range:
        od = base_order_dict.copy()
        od['discount'] = d
        sales_d.append(predict_sales(od))
        prob_d.append(grid_rfc.best_estimator_.predict_proba(pd.DataFrame([od]))[0, 1])

    ax1 = axes[0]
    ax1.plot(discount_range, sales_d, 'o-', color='#2b5876')
    ax1.set_xlabel('Discount')
    ax1.set_ylabel('Doanh thu dự báo (USD)', color='#2b5876')
    ax1.tick_params(axis='y', labelcolor='#2b5876')
    ax1.axvline(base_order_dict['discount'], color='gray', linestyle=':', label='Giá trị đang chọn')
    ax2 = ax1.twinx()
    ax2.plot(discount_range, prob_d, 's--', color='#c0392b')
    ax2.set_ylabel('Xác suất có lãi', color='#c0392b')
    ax2.set_ylim(0, 1)
    ax1.set_title('Biến thiên theo Discount')
    ax1.legend(loc='upper right', fontsize=8)

    ship_range = np.linspace(0, 100, 17)
    sales_s, prob_s = [], []
    for s in ship_range:
        od = base_order_dict.copy()
        od['shipping_cost'] = s
        sales_s.append(predict_sales(od))
        prob_s.append(grid_rfc.best_estimator_.predict_proba(pd.DataFrame([od]))[0, 1])

    ax3 = axes[1]
    ax3.plot(ship_range, sales_s, 'o-', color='#2b5876')
    ax3.set_xlabel('Shipping Cost')
    ax3.set_ylabel('Doanh thu dự báo (USD)', color='#2b5876')
    ax3.tick_params(axis='y', labelcolor='#2b5876')
    ax3.axvline(base_order_dict['shipping_cost'], color='gray', linestyle=':')
    ax4 = ax3.twinx()
    ax4.plot(ship_range, prob_s, 's--', color='#c0392b')
    ax4.set_ylabel('Xác suất có lãi', color='#c0392b')
    ax4.set_ylim(0, 1)
    ax3.set_title('Biến thiên theo Shipping Cost')

    plt.tight_layout()
    plt.show()
```
